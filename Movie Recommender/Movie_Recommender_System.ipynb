```python
import numpy as np
import pandas as pd
import ast
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies = movies.merge(credits,on="title")
```

```python
# genres
# keywords
# id
# title
# overview
# cast
# crew
```

```python
movies = movies[['genres','keywords','id','title','overview','cast','crew']]
```

movies.head(1)

```python
movies.head(1)
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
def convert(obj):
    if isinstance(obj, str):
        obj = ast.literal_eval(obj)
        
    return [i['name'] if isinstance(i, dict) else i for i in obj]
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies['keywords']
```

```python
import ast

def convert3(obj):
    if isinstance(obj, str):
        obj = ast.literal_eval(obj)
        
    return [i['name'] if isinstance(i, dict) else i for i in obj[:3]]
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies['cast']
```

```python
movies.head(1)
```

```python
def fetch_director(obj):
    if isinstance(obj, str):
        obj = ast.literal_eval(obj)
        
    L = []
    for i in obj:
        if isinstance(i, dict) and i.get('job') == 'Director':
            L.append(i['name'])
            break
            
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies['crew']
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head(1)
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_df = movies[['id', 'title' , 'tags']]
```

```python
new_df
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df['tags'][0]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())  
```

```python
new_df
```

```python
from sklearn.feature_extraction.text import CountVectorize
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors[0]
```

```python
list(cv.get_feature_names_out())
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
def stem(text):
    L = []
    for i in text.split():
        L.append(ps.stem(i))

    return " ".join(L)
```

```python
new_df.loc[:, 'tags'] = new_df['tags'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cs = cosine_similarity(vectors)
cs
```

```python
sorted(list(enumerate(cs[0])),reverse = True, key = lambda x:x[1])[1:6]
```

```python
def recommend(movie):
    if movie not in new_df['title'].values:
        print(f"Movie '{movie}' not found in dataset.")
        return
        
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = cs[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```

```python
recommend('Batman Begins')
```

```python
recommend('Inception')
```
